```python
#@title Prueba de conexión
PROYECTO_GEE = "mi-proyecto-gee"  #@param {type:"string"}
import ee
try:
    ee.Initialize(project=PROYECTO_GEE)
except Exception:
    ee.Authenticate()
    ee.Initialize(project=PROYECTO_GEE)
bogota = ee.Geometry.Point([-74.08, 4.65])
ppt = (ee.ImageCollection("UCSB-CHG/CHIRPS/DAILY").filterDate("2020-01-01", "2021-01-01").sum()
       .reduceRegion(ee.Reducer.first(), bogota, 5566).get("precipitation").getInfo())
print(f"Lluvia total 2020 en Bogotá según CHIRPS: {ppt:,.0f} mm")
print("Todo listo ✓ — su proyecto de Earth Engine funciona en Colab.")
```
